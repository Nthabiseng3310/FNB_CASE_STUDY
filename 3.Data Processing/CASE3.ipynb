```sql
SELECT *
 FROM `workspace`.`default`.`fnb3`;

 SELECT DISTINCT
        DATE_FORMAT(DATE, 'MMMM')AS Month_name
        FROM `workspace`.`default`.`fnb3`;

SELECT DISTINCT
        DATE_FORMAT(Date, 'yyyy') AS Year
       FROM `workspace`.`default`.`fnb3`;

SELECT DISTINCT
       DATE_FORMAT(DATE, 'HH:mm:ss')AS Time
       FROM `workspace`.`default`.`fnb3`;


SELECT 
    Date,
    DATE_FORMAT(Date, 'EEEE') AS Day_Name
FROM `workspace`.`default`.`fnb3`;

-- Calculating the daily sales unit price--
SELECT Date,
       Sales,
       Quantity_Sold, ROUND(Sales / Quantity_Sold, 2) AS price_per_unit
       FROM `workspace`.`default`.`fnb3`
       ORDER BY Date;

       SELECT 
       Date,
    DATE(Date) AS Date_Part,
    DATE_FORMAT(Date, 'HH:mm:ss') AS Time
FROM `workspace`.`default`.`fnb3`;

         SELECT Sales,
         ROUND(CAST(Sales AS DOUBLE), 2) AS Sales_Rounded
         FROM `workspace`.`default`.`fnb3`;

         SELECT Cost_Of_Sales,
        ROUND(CAST(Cost_Of_Sales AS DOUBLE),2)AS Cost_Of_Sales_Rounded
        FROM `workspace`.`default`.`fnb3`;

--Calculating the daily Avarage sales unit price of the prroduct--
SELECT
  ROUND(SUM(Sales) / SUM(Quantity_Sold), 2)   AS avg_unit_price_volume_weighted,  
  ROUND(AVG(Sales / Quantity_Sold), 2)        AS avg_unit_price_simple_mean  
  FROM `workspace`.`default`.`fnb3`;

  --Calculating daily percentage gross profit--
  SELECT Date,
         Sales, 
         Cost_Of_Sales,
          ROUND((Sales - Cost_Of_Sales) / Sales * 100, 2) AS Gross_Profit_Percentage
          FROM  `workspace`.`default`.`fnb3`
          ORDER BY Date;
---Calculating the profit/loss--
SELECT Date,
       Sales,
       Cost_Of_Sales,
       ROUND(Sales - Cost_Of_Sales, 2) AS Profit
FROM `workspace`.`default`.`fnb3`
ORDER BY Date;
-- Calculating daily % gross profit per/unit price--
--step 1;calculating the unit cost--
SELECT 
    Date,
    Sales,
    Cost_Of_Sales,
    Quantity_Sold,
    ROUND(Cost_Of_Sales / Quantity_Sold, 2) AS Unit_Cost,
    ROUND(
        ((Sales / Quantity_Sold) - (Cost_Of_Sales / Quantity_Sold)) 
        / (Sales / Quantity_Sold) * 100, 
        2
    ) AS Gross_Profit_Pct_Per_Unit
    FROM `workspace`.`default`.`fnb3`
    ORDER BY Date;

--Promotion periods--


    SELECT
        Date,
        Sales,
        Cost_Of_Sales,
        Quantity_Sold,

        CASE
            WHEN Date BETWEEN '2015-03-24' AND '2015-04-06'
                THEN 'Promotion 1'

            WHEN Date BETWEEN '2016-01-22' AND '2016-02-08'
                THEN 'Promotion 2'

            WHEN Date BETWEEN '2016-08-24' AND '2016-09-06'
                THEN 'Promotion 3'

            ELSE 'Normal'
        END AS Period_Type
       FROM `workspace`.`default`.`fnb3`;

--Calculating the PRICE ELASTICITY DEMAND ON Promotion periods--
CREATE OR REPLACE TABLE `workspace`.`default`.`fnb_final_daily`
USING DELTA AS (
    SELECT 
        CAST(Date AS DATE) AS Date,
        Quantity_Sold,
        
        -- Date/Time dimensions
        DATE_FORMAT(Date, 'HH:mm:ss') AS Time,
        DATE_FORMAT(Date, 'MMMM') AS Month_name,
        DATE_FORMAT(Date, 'yyyy') AS Year,
        DATE_FORMAT(Date, 'EEEE') AS Day_Name,
        

        -- Price calculations
        ROUND(CAST(Sales / Quantity_Sold AS DOUBLE), 2) AS Price_Per_Unit,
        ROUND(CAST(Sales AS DOUBLE), 2) AS Sales_Rounded,
        ROUND(CAST(Cost_Of_Sales AS DOUBLE), 2) AS Cost_Of_Sales_Rounded,
        
        -- Profit calculations
        ROUND(Sales - Cost_Of_Sales, 2) AS Profit,
        ROUND((Sales - Cost_Of_Sales) / Sales * 100, 2) AS Gross_Profit_Percentage,
        
        -- Unit cost and profit per unit
        ROUND(Cost_Of_Sales / Quantity_Sold, 2) AS Unit_Cost,
        ROUND(
            ((Sales / Quantity_Sold) - (Cost_Of_Sales / Quantity_Sold)) 
            / (Sales / Quantity_Sold) * 100, 
            2
        ) AS Gross_Profit_Pct_Per_Unit,


        
        -- Promotion period classification
        CASE 
            WHEN Date BETWEEN '2015-03-24' AND '2015-04-06'
                THEN 'Promotion 1'
            WHEN Date BETWEEN '2016-01-22' AND '2016-02-08'
                THEN 'Promotion 2'
            WHEN Date BETWEEN '2016-08-24' AND '2016-09-06'
                THEN 'Promotion 3'
            ELSE 'Normal'
        END AS Period_Type
    FROM `workspace`.`default`.`fnb3`
);

-- Final table: all rows, no limit
SELECT * FROM `workspace`.`default`.`fnb_final_daily`
ORDER BY Date;

ALTER TABLE `workspace`.`default`.`fnb_final_daily`
SET TBLPROPERTIES ('delta.columnMapping.mode' = 'name');

ALTER TABLE `workspace`.`default`.`fnb_final_daily`
DROP COLUMN `Time`;

SELECT *
FROM `workspace`.`default`.`fnb_final_daily`
ORDER BY DATE;
```
